## BPE + ByteLevel 一句话

> **BPE** 按频率合并高频相邻字节对,学到子词;**ByteLevel** 以 256 个字节为起点,保证任何文本都不会产生 `<unk>`。minimind 用 vocab=6400 的 ByteLevel BPE,中文压缩率约 1.6 chars/token。

## 特殊 token 速查表

| Token | ID | 用途 |
|---|---|---|
| `<\|endoftext\|>` | 0 | 文档分隔 / pad / unk |
| `<\|im_start\|>` | 1 | 消息开始(bos) |
| `<\|im_end\|>` | 2 | 消息结束(eos) |
| `<tool_call>` | ~22 | 函数调用开始 |
| `</tool_call>` | ~23 | 函数调用结束 |
| `<think>` | ~26 | 推理过程开始 |
| `</think>` | ~27 | 推理过程结束 |
| `<\|buffer1-8\|>` | 29-36 | 预留扩展位 |

> 完整 36 个 special token 列表见 `train_tokenizer.py:28-42`。

````python
# 快速验证:加载 tokenizer 查看 special tokens
# 运行前确保已安装: pip install transformers

import os
try:
    from transformers import AutoTokenizer
    tok = AutoTokenizer.from_pretrained("/home/minimind/model")
    print(f"vocab_size: {len(tok)}")
    print(f"special tokens ({len(tok.all_special_tokens)} 个):")
    for i, t in enumerate(tok.all_special_tokens[:10]):
        tid = tok.convert_tokens_to_ids(t)
        print(f"  id={tid:>3}  {t}")
    if len(tok.all_special_tokens) > 10:
        print(f"  ... 共 {len(tok.all_special_tokens)} 个")
except Exception as e:
    print(f"无法加载 tokenizer: {e}")
    print("请确保 /home/minimind/model 目录存在且 transformers 已安装。")
````
