## 知识蒸馏

$$\mathcal{L} = \alpha \cdot CE + (1-\alpha) \cdot T^2 \cdot KL(\text{student}(x/T) \| \text{teacher}(x/T))$$

- T=温度:软化分布,暴露暗知识
- T²缩放:补偿梯度幅度
- α=0.5:平衡硬标签和软标签
- 默认:MoE teacher → dense student

## MoE

- N 个专家 FFN + gate 路由器(hidden → N)
- Top-1 路由:每 token 只激活 1 个专家
- aux_loss = load × scores.mean:防止路由崩塌
- Dense 64M vs MoE 198M(激活 ~64M)

## 关键数字

```
蒸馏: α=0.5, T=1.5, lr=5e-6
MoE:  4 experts, top-1, aux_coef=5e-4
```
