# Agentic RL(精简版)

> [ch14.ipynb](./ch14.ipynb) 的浓缩版。

## 多轮轨迹

```
prompt → model → <tool_call> → execute → <tool_response> → model → final answer → reward
```

整条轨迹共享一个 reward(延迟奖励)。

## 奖励组成

```
reward = tool_validity + 2.5×(verified/gt) + format ± unfinished + rm_score
       ∈ [-3, 3]
```

## 关键设计

- **Mock 工具**:6 个确定性工具,保证可复现
- **response_mask**:只覆盖 assistant 生成,不含 tool_response
- **max_turns=3**:最多 3 轮工具调用
- **gt 弱监督**:只标注期望结果,不标注中间步骤
